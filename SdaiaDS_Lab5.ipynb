```python
storyboard = [

    ["BLUF", "إعطاء الأولوية للاستثمار في التقنيات المستدامة."],

    ["Situation", "الاستثمارات الحالية لا تركز بما يكفي على الاستدامة."],

    ["Complication", "التخلف في الابتكار المستدام يعرض النمو المستقبلي للخطر."],

    ["Evidence", "تظهر اتجاهات السوق تزايد الطلب على الحلول الصديقة للبيئة وتقدم المنافسين."],

    ["Options", "الحفاظ على استراتيجية الاستثمار الحالية أو التحول نحو التكنولوجيا المستدامة."],

    ["Recommendation", "تحويل التركيز الاستثماري نحو التقنيات المستدامة للنمو طويل الأجل."],

    ["Ask", "الموافقة على التحول الاستراتيجي نحو الاستثمار في التكنولوجيا المستدامة."]

]

for slide, message in enumerate(storyboard, start=1):

    print(f"شريحة {slide}: {message[0]}")

    print("الرسالة:", message[1])

    print()
```

```python
def audit(storyboard):

    checks = {

        "سبعة مشاهد بالضبط": len(storyboard) == 7,

        "تبدأ بـ BLUF": storyboard[0][0] == "BLUF",

        "تنتهي بـ Ask": storyboard[-1][0] == "Ask",

        "كل مشهد له رسالة": all(

            message.strip() for _, message in storyboard

        )

    }

    for check, result in checks.items():

        print(f"{'اجتاز' if result else 'فشل'} — {check}")

    return all(checks.values())

audit(storyboard)
```

```python
import pandas as pd

import matplotlib.pyplot as plt

df = pd.DataFrame({

    "نوع التقنية": ["الطاقة الشمسية", "طاقة الرياح", "الطاقة المائية", "تقنيات التدوير"],

    "نسبة الاستثمار": [45, 30, 15, 10]

})

target = 60 # هدف افتراضي لنسبة الاستثمار في إحدى التقنيات الرئيسية أو الإجمالي

print(df)
```

```python
import arabic_reshaper
from bidi.algorithm import get_display

def ar(text):
    reshaped_text = arabic_reshaper.reshape(text)
    return get_display(reshaped_text)

plt.figure(figsize=(10, 6))

# Apply ar function to x-axis labels and plot
x_labels_reshaped = [ar(label) for label in df["نوع التقنية"]]
bars = plt.bar(x_labels_reshaped, df["نسبة الاستثمار"], color='lightgreen')

# Add numerical labels on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1, round(yval, 2), ha='center', va='bottom') # +1 to place text slightly above the bar

plt.title(ar("نسبة الاستثمار في أنواع التقنيات المستدامة"), fontsize=16)
plt.xlabel(ar("نوع التقنية"), fontsize=12)
plt.ylabel(ar("نسبة الاستثمار (%)"), fontsize=12)

# Set x-tick labels with the reshaped text
plt.xticks(rotation=45, ha='right') # Rotate for better readability if labels are long

plt.ylim(0, 100)

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
import arabic_reshaper
from bidi.algorithm import get_display

def ar(text):
    reshaped_text = arabic_reshaper.reshape(text)
    return get_display(reshaped_text)

plt.figure(figsize=(10, 6))

x_labels_reshaped = [ar(label) for label in df["نوع التقنية"]]
bars = plt.bar(x_labels_reshaped, df["نسبة الاستثمار"], color='lightgreen')

# Add numerical labels on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1, round(yval, 2), ha='center', va='bottom') # +1 to place text slightly above the bar

plt.axhline(

    target,

    linestyle="--",

    color='red',

    label=ar(f"الهدف = {target}%")

)

plt.title(ar("نسبة الاستثمار الحالية مقابل الهدف"), fontsize=16)

plt.xlabel(ar("نوع التقنية"), fontsize=12)

plt.ylabel(ar("نسبة الاستثمار (%)"), fontsize=12)

plt.ylim(0, 100)

plt.xticks(rotation=45, ha='right')

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.legend()

plt.tight_layout()

plt.show()
```

```python
import arabic_reshaper
from bidi.algorithm import get_display

def ar(text):
    reshaped_text = arabic_reshaper.reshape(text)
    return get_display(reshaped_text)

df["Gap"] = target - df["نسبة الاستثمار"]

print(df[["نوع التقنية", "نسبة الاستثمار", "Gap"]])

plt.figure(figsize=(10, 6))

x_labels_reshaped = [ar(label) for label in df["نوع التقنية"]]
bars = plt.bar(x_labels_reshaped, df["Gap"], color='lightcoral')

# Add numerical labels on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1, round(yval, 2), ha='center', va='bottom') # +1 to place text slightly above the bar

plt.axhline(

    0,

    linestyle="--",

    color='black'

)

plt.title(ar("الفجوة عن هدف الاستثمار في التقنيات المستدامة"), fontsize=16)

plt.xlabel(ar("نوع التقنية"), fontsize=12)

plt.ylabel(ar("الفجوة عن الهدف (نقطة مئوية)"), fontsize=12)

plt.ylim(0, df["Gap"].max() * 1.2) # Adjust y-limit dynamically

plt.xticks(rotation=45, ha='right')

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()

plt.show()
```
