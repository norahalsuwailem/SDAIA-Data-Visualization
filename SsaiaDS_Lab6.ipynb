```python
# 1. تعريف الكلاس الخاص بالشرائح لتوفير خصائص arc_role و message
class Slide:
    def __init__(self, arc_role, message):
        self.arc_role = arc_role
        self.message = message

# 2. تحديد الفكرة الرئيسية
BIG_IDEA = "Improve IT System Uptime to Reduce Departmental Absence"

# 3. بناء القصة كاملة بأدوارها
STORYBOARD = [
    Slide("situation", "IT team faces high workload and frequent manual fixes."),
    Slide("complication", "Absence rate in IT reached 12%, higher than all other departments."),
    Slide("evidence", "Data shows 60% of absence correlates with system outage alerts."),
    Slide("options", "Implement automated system monitoring or hire additional support staff."),
    Slide("resolution", "Deploy automated fault prediction to reduce stress and downtime.")
]

TIME_BOX = {
    "BLUF": 30,
    "situation": 45,
    "complication": 60,
    "evidence": 75,
    "options": 90,
    "resolution": 60
}

# 4. الدالة الأصلية
def build_notes(storyboard):
    total = 0
    print(f"OPEN (say the ask): {BIG_IDEA}\n")
    for s in storyboard:
        secs = TIME_BOX.get(s.arc_role, 45)
        total += secs
        print(f"[{total//60}:{total%60:02d}] {s.arc_role}")
        print(f"   Orient axes, then: {s.message}")
        print("   Point at one mark; do NOT read it.")

    assert total <= 7*60, f"Over by {total-420}s – cut."
    print("\nCLOSE (ask AGAIN): approve; 90-day check.")

# 5. التشغيل
build_notes(STORYBOARD)
```

```python
QA = [
 {"q": "Aren't those regions just small? (hostile)",
  "answer": "Per-capita, not absolute (appx A2).",
  "bridge": "so targeted spend closes the real gap."},

 {"q": "Adoption looks high vs finance (data-wrong)",
  "answer": "Unique users, not a column avg (A5).",
  "bridge": "it holds at finance's lower figure."},

 {"q": "Why not spend it on the app? (off-topic)",
  "answer": "Channel infra is option 2; slower.",
  "bridge": "region-targeting is fastest to 65%."},
]

def drill(qa):
   for i, item in enumerate(qa, 1):
       print(f"Q{i}: {item['q']}")
       print(f"  A: {item['answer']}")
       print(f"  ->bridge: ...{item['bridge']}")

drill(QA)
```
