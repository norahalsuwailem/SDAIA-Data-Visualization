```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
   "Department": ["HR", "Finance", "IT", "Sales"],
   "Absence": [5, 8, 12, 7]
})

plt.bar(df["Department"], df["Absence"],
       color=["red", "green", "blue", "purple"])
plt.title("Employee Absence")
plt.xlabel("Department")
plt.ylabel("Absence %")
plt.grid(True)
plt.legend(["Absence"])
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("IT Has the Highest Absence — 12%")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.annotate("Highest",
            xy=("IT", 12),
            xytext=("Finance", 13),
            arrowprops=dict(arrowstyle="->"))
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")

plt.ylim(0, 15)

plt.title("IT Has the Highest Absence — 12%", pad=15)

plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.annotate("Highest",
            xy=("IT", 12),
            xytext=("Finance", 13),
            arrowprops=dict(arrowstyle="->"))

for i, v in enumerate(df["Absence"]):
    plt.text(i, v + 0.5, str(v) + "%", ha='center', va='bottom')

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.show()
```

```python
arabic = ["الموارد البشرية", "المالية", "تقنية المعلومات", "المبيعات"]

plt.figure(figsize=(9,5))
plt.bar(arabic, df["Absence"], color="steelblue")
plt.title("تقنية المعلومات لديها أعلى نسبة غياب — 12%")
plt.xlabel("القسم")
plt.ylabel("نسبة الغياب (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import arabic_reshaper
from bidi.algorithm import get_display

def ar(text):
    reshaped_text = arabic_reshaper.reshape(text)
    return get_display(reshaped_text)

df = pd.DataFrame({
    "Department": [ar("الموارد البشرية"), ar("المالية"), ar("تقنية المعلومات"), ar("المبيعات")],
    "Absence": [5, 8, 12, 7]
})

plt.figure(figsize=(8, 5))
plt.bar(df["Department"], df["Absence"], color="green")

plt.ylim(0, 15)
plt.title(ar("تقنية المعلومات لديها أعلى نسبة غياب — 12%"), pad=15)
plt.xlabel(ar("القسم"))
plt.ylabel(ar("نسبة الغياب (%)"))
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.show()
```
